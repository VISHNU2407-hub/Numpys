```python
# NumPy - Multi-Dimensional Indexing
#array[layer,row,column]
import numpy as np

# Creating a 3D array
array = np.array([
    [
        ['A', 'B'],
        ['C', 'D']
    ],
    [
        ['M', 'N'],
        ['O', 'P']
    ]
])

print("3D Array:")
print(array)

# Shape and dimensions
print("\nDimensions:", array.ndim)
print("Shape:", array.shape)

# Multi-dimensional indexing
print("\nIndexing Examples:")

print("array[0, 0, 0] =", array[0, 0, 0])  # A
print("array[0, 0, 1] =", array[0, 0, 1])  # B
print("array[0, 1, 0] =", array[0, 1, 0])  # C
print("array[0, 1, 1] =", array[0, 1, 1])  # D

print("array[1, 0, 0] =", array[1, 0, 0])  # M
print("array[1, 0, 1] =", array[1, 0, 1])  # N
print("array[1, 1, 0] =", array[1, 1, 0])  # O
print("array[1, 1, 1] =", array[1, 1, 1])  # P

# Building a word using indexing
word = array[0, 0, 0] + array[1, 0, 0] + array[1, 0, 1]

print("\nWord:", word)
```
