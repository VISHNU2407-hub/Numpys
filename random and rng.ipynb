```python

# ============================================================
# RANDOM NUMBERS IN NUMPY
# ============================================================

# ------------------------------------------------------------
# 1. Seeding for Reproducibility
# ------------------------------------------------------------
# Same seed → same random numbers every time
# Similar to a Minecraft world seed:
# Same seed → same world
#
# default_rng() is the recommended modern approach.

rng = np.random.default_rng(seed=1)

print("Random numbers:")
print(rng.random(5))


```

```python
# ------------------------------------------------------------
# 2. Uniform Random Floats
# ------------------------------------------------------------
# rng.uniform() → random float between 0 and 1
# Range: [0, 1)
#
# low = starting value
# high = ending value

print("\nRandom float (0 to 1):")
print(rng.uniform())

print("\nRandom float (-1 to 1):")
print(rng.uniform(low=-1, high=1))


# Generate 3 random floats
print("\n3 Random floats:")
print(rng.uniform(low=-1, high=1, size=3))


# Generate a 3 × 2 array
print("\n3 × 2 Random array:")
print(rng.uniform(low=-1, high=1, size=(3, 2)))


```

```python
# ------------------------------------------------------------
# 3. Shuffling
# ------------------------------------------------------------
# shuffle() randomly reorders an array IN PLACE
# → Original array itself gets changed

numbers = np.array([1, 2, 3, 4, 5])

print("\nBefore shuffle:")
print(numbers)

rng.shuffle(numbers)

print("After shuffle:")
print(numbers)


# For a 2D array:
# shuffle() reorders rows (first axis)

matrix = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print("\nBefore 2D shuffle:")
print(matrix)

rng.shuffle(matrix)

print("After 2D shuffle:")
print(matrix)


```

```python
# ------------------------------------------------------------
# 4. Random Choice
# ------------------------------------------------------------
# choice() randomly selects elements from an array
#
# replace=True  → element can be selected again
# replace=False → once selected, cannot be selected again

fruits = np.array([
    "apple",
    "orange",
    "banana",
    "coconut",
    "pineapple"
])

print("\nRandom fruit:")
print(rng.choice(fruits))


# Select 3 fruits
print("\n3 Random fruits (replace=True):")
print(rng.choice(fruits, size=3, replace=True))

#replace=True =>can select same element again and again
#replace=False=>can't select same element twice
# Select 3 different fruits
print("\n3 Random fruits (replace=False):")
print(rng.choice(fruits, size=3, replace=False))


```

```python
# ------------------------------------------------------------
# 5. 2D Random Choice
# ------------------------------------------------------------
# size=(3,3) → creates a 3 × 3 array

print("\n3 × 3 Random fruits:")
print(rng.choice(fruits, size=(3, 3)))


```

```python
# ============================================================
# QUICK REVISION
# ============================================================
# np.random.default_rng(seed)
# → creates a random number generator
#
# Same seed → same random sequence
#
# rng.uniform(low, high, size)
# → random floating-point numbers
#
# rng.shuffle(array)
# → randomly reorders the array IN PLACE
# → 2D: shuffles rows
#
# rng.choice(array, size, replace)
# → randomly selects elements
#
# replace=True
# → repetition allowed
#
# replace=False
# → no repetition
#
# Modern approach:
# np.random.default_rng()
#
# Older approach:
# np.random.seed()
# ============================================================
```
