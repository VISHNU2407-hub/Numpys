```python
import numpy as np

# Array of floats
array = np.array([1.01, 2.5, 3.99])

print("Array:", array)

print("Square root:", np.sqrt(array))
print("Round:", np.round(array))
print("Floor:", np.floor(array))
print("Ceil:", np.ceil(array))

# Mathematical constant
print("Pi:", np.pi)

# Circle areas
radii = np.array([1, 2, 3])

areas = np.pi * radii ** 2

print("Radii:", radii)
print("Areas:", areas)

# Scalar arithmetic
print("Multiply by 3:", radii * 3)

# Vectorized exponentiation
print("Squared:", radii ** 2)

# Vectorized function
print("Square root:", np.sqrt(radii))
```
