```python
import numpy as np

# Elementwise Arithmetic Between Arrays

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print("Array 1:", array1)
print("Array 2:", array2)

print("Addition:", array1 + array2)
print("Subtraction:", array1 - array2)
print("Multiplication:", array1 * array2)
print("Division:", array1 / array2)
print("Power:", array1 ** array2)


# Comparison Operators

scores = np.array([100, 85, 45, 60, 72, 55, 100])

print("\nScores:", scores)

print("Scores == 100:", scores == 100)
print("Scores >= 60:", scores >= 60)
print("Scores < 60:", scores < 60)


# Boolean Masking

print("\nFailing Scores:")
print(scores[scores < 60])


# Conditional Assignment

scores[scores < 60] = 0

print("\nScores after replacing failing scores with 0:")
print(scores)
```
