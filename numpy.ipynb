```python
import numpy as np
```

```python
#Arrays
#Creating an array in numpy
array=np.array([1,2,3,4])
print(type(array))
```

```python
#scalar operation on array
s=array*2 #multiples each element by 2
print(s)
t=array+1 # increment each element by 1
print(t)
u=array-2 #decrement each element by 2
print(u)
v=array/4 #float division
print(v)
array//4  #interger division
```

```python
#duplicating the array
arr=np.array([1,2,3,4])
print([1,2,3,4]*2 )#duplciates the element rather than multiplying
[1,2,3,4]*3
```

```python
arr = np.array([10, 20, 30])
# dtype for the data type of the array
print(arr.dtype)
```

```python
#dimensions
#ndim reports number of dimensions
```

```python


# 1D Array
arr1 = np.array([1, 2, 3])

# 2D Array
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D Array
arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("ndim:", arr1.ndim)

print("ndim:", arr2.ndim)

print("ndim:", arr3.ndim)
```

```python
#shape
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.shape)
```

```python
#value error ,unequal lengths of rows
arr = np.array([
    [1, 2, 3],
    [4, 5]
])

print(arr)
print(arr.shape)
```
